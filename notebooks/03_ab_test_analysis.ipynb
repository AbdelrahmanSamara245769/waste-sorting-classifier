# A/B Test Analysis

Compares two versions of the Smart Bin app prototype: version A has a scan history, version B a feedback button. Each version was rated by 10 participants on 7-point Likert items, and an independent-samples t-test compares the means.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

Load the survey responses for both versions. The responses aren't included in this repository.

```python
# Survey responses, one row per participant
df_A = pd.read_excel("../data/ab_test_version_a.xlsx")
df_B = pd.read_excel("../data/ab_test_version_b.xlsx")

# EDA A
df_A.info()
df_A.head()

# EDA B
df_B.info()
df_B.head()
```

A t-test assumes that each item is roughly normally distributed in both groups (Shapiro–Wilk) and that both groups have equal variances (Levene). Both assumptions are checked per question before testing.

### Test Assumptions

```python
df_A = df_A.rename(columns={
    "I understood what I could use the app for": "I understood what I could use the app for."
})

questions = [
    "I understood what I could use the app for.",
    "I found the application intuitive to use.",
    "I thought the application was useful.",
    "I enjoyed using the application."
]

for q in questions:
    # Run the shapiro-wilk statistical test for each question to check whether the data is normally distributed
    normal_a = stats.shapiro(df_A[q])
    normal_b = stats.shapiro(df_B[q])

    # Check whether the variance of both samples is equal
    homogeneity = stats.levene(df_A[q],
                            df_B[q])

    print("These are the results for question " + q)
    print(f"This is the p-value for normality for version A: ", normal_a)
    print(f"This is the p-value for normality for version B: ", normal_b)
    print("A p-value above 0.05 means the data is normally distributed. If for either version A or B the data is not normally distributed, you will have to run the bootstrapped version for this question.")
    print(f"This is the p-value for homogeneity: ", homogeneity)
    print("If the p-value is above 0.05, then the groups have equal variances. If the variance aren't equal then you will have to run the bootstrapped version for this question.")
```

### Run the t-test

The assumptions hold for "I thought the application was useful." (Shapiro–Wilk p = 0.12 and 0.11, Levene p = 0.48), so the regular independent-samples t-test is used.

```python
# Run Independent Samples T-test when assumptions are not violated.
results = stats.ttest_ind(df_A["I thought the application was useful."],
                          df_B["I thought the application was useful."])

# Print the results
print(f"The results are significant if the p-value is significant, which means smaller than 0.05", 
results)
```

**Result:** t(18) = 1.23, p = 0.23. There is no significant difference in perceived usefulness between the two versions, so the feedback button alone did not change how useful participants found the app.
